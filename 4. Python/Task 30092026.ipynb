```python
## Safe Data Type Conversion

class MockDataFile:
    def __init__(self, file_name):
        self.file_name = file_name
        self.is_open = False

    def open(self):
        self.is_open = True
        print(f"Opened file: {self.file_name}")

    def read(self):
        if not self.is_open:
            raise RuntimeError("Cannot read: file is not open")

        print("Reading file data...")
        return ["100", "200", "ERROR", "400"]

    def close(self):
        self.is_open = False
        print(f"Closed file: {self.file_name}")


def process_data_file():
    data_file = MockDataFile("customer_data.csv")
    data_file.open()

    try:
        rows = data_file.read()

        for row in rows:
            if row == "ERROR":
                raise ValueError("Invalid value found while processing row")

            print(f"Processed row: {row}")

    except ValueError as error:
        print(f"Processing failed: {error}")

    finally:
        data_file.close()
        print(f"File open status after cleanup: {data_file.is_open}")


process_data_file()
```

```python
##Resource cleanup simulation

def convert_to_integers(raw_values):
    converted_values = []
    rejected_values = []

    for value in raw_values:
        try:
            number = int(value)
            converted_values.append(number)
        except (ValueError,TypeError):
            converted_values.append(None)
            rejected_values.append(value)
        
    return converted_values, rejected_values

raw_data = ["10", "002", "abc", "30", None, "40.5", "-50", "60", "70", "xyz", "80", "90", "100" , "Ling"]

cleaned_data, invalid_data = convert_to_integers(raw_data)

print("Converted values:", cleaned_data)
print("Rejected values:", invalid_data)
```

```python
##Custom Data Validation

class InvalidAgeError(Exception):
    """Raised when customer age is outside the allowed range."""
    pass


def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError(
            f"Suspicious age: {age}. Allowed range is 0 to 120."
        )

    return age


customer_ages = [25, 40, -5, 130, 0, 120, 35]

valid_ages = []
invalid_ages = []

for age in customer_ages:
    try:
        valid_age = validate_age(age)
        valid_ages.append(valid_age)

    except InvalidAgeError as error:
        print(error)
        invalid_ages.append(age)

print("Valid ages:", valid_ages)
print("Invalid ages:", invalid_ages)
```
