```python
#Inventory Filtering System 

stock_levels = [12, 45, 22, 60, 33, 88]

# Keep only stock levels divisible by 2
even_stock_levels = []

for stock in stock_levels:
    if stock % 2 == 0:
        even_stock_levels.append(stock)

print("All stock levels:", stock_levels)
print("Stocks requiring warehouse binning:", even_stock_levels)
```

```python
# Memory Footprint Simulation

#List
inventory_list = [12, 45, 22]

list_id_before = id(inventory_list)

inventory_list = [60, 33, 88]

list_id_after = id(inventory_list)

print("List ID before:", list_id_before)
print("List ID after :", list_id_after)
print("List object changed:", list_id_before != list_id_after)


# Tuple
inventory_tuple = (12, 45, 22)

tuple_id_before = id(inventory_tuple)

inventory_tuple = (60, 33, 88)

tuple_id_after = id(inventory_tuple)

print("\nTuple ID before:", tuple_id_before)
print("Tuple ID after :", tuple_id_after)
print("Tuple object changed:", tuple_id_before != tuple_id_after)
```

```python
#Advanced Analytics Pipeline

def process_data(raw_data):
    # Step 1: Remove outliers greater than 100
    without_outliers = [
        value for value in raw_data
        if value <= 100
    ]

    # Step 2: Keep only even numbers
    even_values = [
        value for value in without_outliers
        if value % 2 == 0
    ]

    # Step 3: Convert final result into an immutable tuple
    final_report = tuple(even_values)

    return final_report


raw_data = [12, 45, 22, 150, 60, 101, 33, 88, 120, 90, 99, 4]

result = process_data(raw_data)

print("Raw data:", raw_data)
print("Final immutable report:", result)
print("Result type:", type(result))
```
