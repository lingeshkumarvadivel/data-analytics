```python
#Year Validator Utility
def is_leap_year(year):
        return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)


try:
    year = int(input("Enter a year: "))

    if is_leap_year(year):
        print(True)
        print(f"{year} is a leap year.")
    else:
        print(False)
        print(f"{year} is not a leap year.")

except ValueError:
    print("Please enter a valid numeric year.")
```

```python
#Robust Date-to-Season Converter
from datetime import datetime


def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)


def date_to_season(date_text):
    seasons = {
        "Winter": [12, 1, 2],
        "Spring": [3, 4, 5],
        "Summer": [6, 7, 8],
        "Autumn": [9, 10, 11]
    }

    try:
        day, month, year = map(int, date_text.strip().split("/"))

        if month == 2 and day == 29 and not is_leap_year(year):
            raise ValueError("Invalid leap-year date.")

        datetime.strptime(date_text.strip(), "%d/%m/%Y")

        for season, months in seasons.items():
            if month in months:
                return season

    except ValueError:
        return "Invalid date. Use DD/MM/YYYY format."


date_input = input("Enter date (DD/MM/YYYY): ")
print(date_to_season(date_input))
```
