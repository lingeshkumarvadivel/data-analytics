```python
#Topic: DataFrame Inspection Exploration
#1. Load a CSV file containing employee records and perform a series of inspection tasks to understand the data structure, data types, and potential missing values.

import pandas as pd

# =========================
# 1. Load the CSV file
# =========================
# Replace "employees.csv" with your actual filename if different
df = pd.read_csv("employees.csv")

# =========================
# 2. Basic shape and preview
# =========================
print("=== BASIC INFO ===")
print("Shape (rows, columns):", df.shape)

print("\nFirst 10 rows:")
print(df.head(10))

print("\nLast 5 rows:")
print(df.tail())

# =========================
# 3. Column names and data types
# =========================
print("\n=== COLUMN NAMES AND DTYPES ===")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

# More detailed info (non-null counts + dtypes)
print("\nDetailed info:")
print(df.info())

# =========================
# 4. Missing values inspection
# =========================
print("\n=== MISSING VALUES ===")
print("Missing values per column:")
print(df.isna().sum())

print("\nPercentage missing per column:")
print((df.isna().sum() / len(df) * 100).round(2))

# Optional: rows with any missing value
rows_with_missing = df[df.isna().any(axis=1)]
print("\nNumber of rows with at least one missing value:", len(rows_with_missing))

# =========================
# 5. Basic statistics for numeric columns
# =========================
print("\n=== SUMMARY STATISTICS (NUMERIC) ===")
print(df.describe())

# =========================
# 6. Unique values for categorical columns (optional but useful)
# =========================
print("\n=== UNIQUE VALUES FOR OBJECT COLUMNS ===")
for col in df.select_dtypes(include=["object", "category"]).columns:
    print(f"\n{col}:")
    print("  Unique count:", df[col].nunique())
    print("  Sample values:", df[col].dropna().unique()[:10])
```
