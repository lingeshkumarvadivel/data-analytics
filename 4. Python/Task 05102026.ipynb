```python
#Topic: Data Structure Distinction: Series vs. DataFrame
#1. In this task, you will create a simple sales dataset. You will define a Series representing product prices and a DataFrame representing a full inventory list.

import pandas as pd

# 1. Series: product prices (index = product names)
prices = pd.Series(
    [120, 250, 80, 300, 150],
    index=["Pen", "Notebook", "Eraser", "Bag", "Lunch Box"],
    name="Price"
)

print("Prices Series:")
print(prices)

# 2. DataFrame: full inventory list
# Columns: Product, Category, Stock, Price
inventory = pd.DataFrame({
    "Product": ["Pen", "Notebook", "Eraser", "Bag", "Lunch Box"],
    "Category": ["Stationery", "Stationery", "Stationery", "Accessories", "Accessories"],
    "Stock": [500, 200, 800, 50, 120],
    "Price": prices  # directly use the Series; index aligns by product name
})

print("\nInventory DataFrame:")
print(inventory)

# Optional: compute total value per product
inventory["TotalValue"] = inventory["Stock"] * inventory["Price"]
print("\nInventory with Total Value:")
print(inventory)
```

```python
#Topic: Mutation vs. Inspection Logic
#3. Perform operations on a DataFrame to distinguish between destructive changes and non-destructive analysis.

import pandas as pd

# Load data
df = pd.read_csv("employees.csv")

# Keep an original copy to compare later
df_original = df.copy()

# =========================
# 1) Non-destructive analysis
# =========================
# These operations do NOT change df; they return new objects or views.

# a) Compute average salary by department (returns a new Series)
avg_salary_by_dept = df.groupby("department")["salary"].mean()

# b) Filter high earners (returns a new DataFrame)
high_earners = df[df["salary"] > 80000]

# c) Add a computed column temporarily (using assign, returns new DataFrame)
df_with_bonus = df.assign(bonus = df["salary"] * 0.1)

print("Original df unchanged after non-destructive ops:")
print(df.head())
print("\nNew DataFrame with bonus column:")
print(df_with_bonus.head())

# =========================
# 2) Destructive changes
# =========================
# These operations MODIFY df in place.

# a) Overwrite a column
df["salary"] = df["salary"] * 1.1  # 10% hike, changes df permanently

# b) Drop a column
df.drop(columns=["is_full_time"], inplace=True)

# c) Filter rows and assign back (loses original rows)
df = df[df["age"] >= 25].reset_index(drop=True)

# d) Fill missing values in place (if any)
# df.fillna({"age": df["age"].median()}, inplace=True)

print("\nDataFrame after destructive changes:")
print(df.head())

print("\nOriginal DataFrame (unchanged):")
print(df_original.head())
```

```python
#Topic: Data Cleansing and Reporting	
#4. You are given a messy sales dataset. You must load it, identify outliers using summary tools, and clean the data for a report.

import pandas as pd

# Load your CSV file
df = pd.read_csv("messy_sales.csv")

# Display original data information
print("Original shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nSummary statistics:")
print(df.describe())

# Create a separate copy for cleaning
df_clean = df.copy()

# Convert date column to datetime
df_clean["date"] = pd.to_datetime(df_clean["date"], errors="coerce")

# Convert numeric columns to numbers
df_clean["quantity"] = pd.to_numeric(df_clean["quantity"], errors="coerce")
df_clean["unit_price"] = pd.to_numeric(df_clean["unit_price"], errors="coerce")

# Identify invalid values
invalid_quantity = df_clean["quantity"] <= 0
invalid_price = df_clean["unit_price"] <= 0

print("\nInvalid quantity rows:", invalid_quantity.sum())
print("Invalid price rows:", invalid_price.sum())

# Remove invalid quantity and price rows
df_clean = df_clean[
    (df_clean["quantity"] > 0) &
    (df_clean["unit_price"] > 0)
].copy()

# Identify quantity outliers using the IQR method
q1_quantity = df_clean["quantity"].quantile(0.25)
q3_quantity = df_clean["quantity"].quantile(0.75)
iqr_quantity = q3_quantity - q1_quantity

quantity_lower_limit = q1_quantity - 1.5 * iqr_quantity
quantity_upper_limit = q3_quantity + 1.5 * iqr_quantity

quantity_outliers = (
    (df_clean["quantity"] < quantity_lower_limit) |
    (df_clean["quantity"] > quantity_upper_limit)
)

print("Quantity outlier rows:", quantity_outliers.sum())

# Identify price outliers using the IQR method
q1_price = df_clean["unit_price"].quantile(0.25)
q3_price = df_clean["unit_price"].quantile(0.75)
iqr_price = q3_price - q1_price

price_lower_limit = q1_price - 1.5 * iqr_price
price_upper_limit = q3_price + 1.5 * iqr_price

price_outliers = (
    (df_clean["unit_price"] < price_lower_limit) |
    (df_clean["unit_price"] > price_upper_limit)
)

print("Price outlier rows:", price_outliers.sum())

# Remove outlier rows
df_clean = df_clean[
    ~quantity_outliers & ~price_outliers
].copy()

# Fill missing values
df_clean["region"] = df_clean["region"].fillna("Unknown")
df_clean["customer_id"] = df_clean["customer_id"].fillna("Unknown")

# Add revenue column
df_clean["revenue"] = df_clean["quantity"] * df_clean["unit_price"]

# Reset row numbering
df_clean.reset_index(drop=True, inplace=True)

# Display cleaned data
print("\nCleaned shape:", df_clean.shape)

print("\nCleaned data:")
print(df_clean.head())

print("\nMissing values after cleaning:")
print(df_clean.isna().sum())

print("\nCleaned summary statistics:")
print(df_clean.describe())

# Save cleaned data for reporting
df_clean.to_csv("cleaned_sales.csv", index=False)

print("\nCleaned file saved as cleaned_sales.csv")
```

```python
#Topic: Advanced Data Construction & Manipulation
#5. You are provided with a complex, nested JSON-like dictionary. You must transform it into a high-performance DataFrame and apply a transformation logic that mutates columns based on conditional rules.

import json
import pandas as pd
import numpy as np

# =========================
# 1. Load the nested JSON file
# =========================
with open("orders.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# =========================
# 2. Flatten nested JSON into a DataFrame (one row per item)
# =========================
rows = []

for order in data["orders"]:
    order_id = order["order_id"]
    cust = order["customer"]
    meta = order["meta"]

    for item in order["items"]:
        row = {
            "order_id": order_id,
            "customer_id": cust["id"],
            "customer_name": cust["name"],
            "customer_segment": cust["segment"],
            "contact_email": cust["contact"]["email"],
            "contact_phone": cust["contact"]["phone"],
            "preferred_channel": cust["contact"]["preferred_channel"],
            "product": item["product"],
            "qty": item["qty"],
            "unit_price": item["unit_price"],
            "discount": item["discount"],
            "region": meta["region"],
            "channel": meta["channel"],
            "is_express": meta["is_express"],
            "notes": meta["notes"]
        }
        rows.append(row)

df = pd.DataFrame(rows)

print("=== RAW DATAFRAME ===")
print(df)

# =========================
# 3. Apply transformation logic that mutates columns
# =========================

# Ensure numeric types
df["qty"] = pd.to_numeric(df["qty"], errors="coerce")
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
df["discount"] = pd.to_numeric(df["discount"], errors="coerce")

# Compute base line_amount
df["line_amount"] = df["qty"] * df["unit_price"]

# Apply discount
df["discounted_amount"] = df["line_amount"] * (1 - df["discount"])

# --- Conditional mutations ---

# 1) Premium customers get an extra 5% discount on discounted_amount
premium_mask = df["customer_segment"] == "Premium"
df.loc[premium_mask, "discounted_amount"] *= 0.95

# 2) Express orders get a flat ₹50 surcharge added
express_mask = df["is_express"] == True
df.loc[express_mask, "discounted_amount"] += 50

# 3) Cap negative or zero amounts to a small positive value
df.loc[df["discounted_amount"] <= 0, "discounted_amount"] = 1.0

# 4) Normalize channel: "Online" -> "Digital", "Offline" -> "Store"
df.loc[df["channel"] == "Online", "channel"] = "Digital"
df.loc[df["channel"] == "Offline", "channel"] = "Store"

# 5) Fill missing phone with "Not provided"
df["contact_phone"] = df["contact_phone"].fillna("Not provided")

# 6) Fill missing notes with "No special instructions"
df["notes"] = df["notes"].fillna("No special instructions")

# 7) Create a priority score based on segment and express
# Premium + Express = 3, Premium only = 2, Express only = 2, else = 1
df["priority_score"] = 1
df.loc[premium_mask, "priority_score"] = 2
df.loc[express_mask, "priority_score"] = 2
df.loc[premium_mask & express_mask, "priority_score"] = 3

print("\n=== TRANSFORMED DATAFRAME ===")
print(df)

print("\nColumn dtypes:")
print(df.dtypes)

# Optionally save to CSV for reporting / Power BI
df.to_csv("transformed_orders.csv", index=False)
print("\nTransformed data saved to transformed_orders.csv")
```
